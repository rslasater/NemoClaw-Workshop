### Start the OpenShell terminal

Openshell provides the framework for managing the permissions of the claw. We will now open a terminal that will allow for the managment of permissions.

In the new terminal enter:
openshell term

This will open up the terminal. Use tab and enter to open up the configuration for the claw that you are currently running.
Type 'r' to open up the current permissions. Note that there are no prebuilt rules allowing the claw to interact outside of its shell.

### Connect the claw to your email server

First determine the ip address of your server. In a terminal run:

hostname -I 

Record the first set of IP addresses.

We need to establish a connection with the email server running on 10.x.x.x:8000
First, go back to your claw terminal and ask it to see if it can see whats running on that port. The claw may respond that it can see a server running, or it may state that it cannot access the port. Next ask the claw to connect to the port. Depending on the claws randomness it may attempt to connect to the port and receive an error that it cannot connect, or it may state that it cannot connect and not attempt to connect. Try to coerce it to connect to the port. Ideally it will try to make the connection and be denied.

IF IT SUCCESSFULLY ATTEMPTS AND IS DENIED THEN PROCEED TO THE NEXT STEP

If it will not attempt to connect on its own, exit out of the tui by entering /exit and we will attempt to reach the server on our own.

Within the shell type in:
curl 10.0.0.0:8000 
You should receive an error that the connection was blocked.

### Grant permission

Open up the terminal that is running the openshell terminal. Under rules you should now see a rule for port 8000. Approve the rule and return to NemoClaw.

Back in your claw tui ask it to connect to the port. It should now be able to access the email server!!!
